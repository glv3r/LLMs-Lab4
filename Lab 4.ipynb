# Section 1 — Talking to an LLM Programmatically

Before building anything, understand the anatomy of an API call: **messages and roles**
(`system`, `user`, `assistant`), and the **generation parameters** (`temperature`,
`max_tokens`).

### Part 1.1 — Your first API call

````python
# TODO: Write a helper function you will reuse for the WHOLE lab:
#
def ask_llm(user_prompt, system_prompt="You are a helpful assistant.", temperature=0.7, max_tokens=500):
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user",   "content": user_prompt},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )

    return response
#
# TODO: Call it once with a simple question and print the answer.
# TODO: Print response.usage as well — how many tokens did your call consume?

res = ask_llm(user_prompt="Which model are you?")
print(f"Response: {res.choices[0].message.content}")
print(f"Usage: {res.usage.completion_tokens} tokens for completion")
````

### Part 3.2 — Component 2: Structured extraction (JSON)
Downstream software cannot read prose. Extract the fields in `GOLD` as strict JSON.

````python
# TODO: Write EXTRACT_PROMPT — a template that instructs the model to return ONLY a JSON
#   object with EXACTLY these keys:
#     applicant_name (string), amount_ghs (number), purpose (string),
#     monthly_profit_ghs (number or null), has_collateral_or_guarantor (boolean),
#     repayment_months (number or null)
#   Techniques to use:
#     - explicit schema in the prompt
#     - ONE worked example (few-shot) using a letter you write yourself (not from LETTERS!)
#     - "If a field is not stated in the letter, use null. Do not guess."
#     - temperature=0


EXTRACT_PROMPT = "You are to carefully examine the prompt the user has given and return ONLY a JSON object with exactly these keys:" \
"applicant_name: string, amount_ghs: number, purpose: string, monthly_profit_ghs: number or null, " \
"has_collateral_or_guarantor: boolean, and repayment_months: number or null. If a field is not stated in the letter, use " \
"null as instructed. DO NOT GUESS."

# TODO: Write extract_fields(letter_text) that calls the LLM, strips any ```json fences,
#   json.loads() the result, and returns a dict. Handle parse failures gracefully
#   (return None and print a warning).

def extract_fields(letter_text, temperature=0.0):
    r = ask_llm(letter_text, EXTRACT_PROMPT, temperature=temperature).choices[0].message.content

    # finding the first and last bracket of a JSON object
    start = r.find("{")
    end = r.rfind("}")

    if start == -1 or end == -1 or end < start:
        print(f"No JSON object found in response")
        return None

    try:
        return json.loads(r[start:end + 1])
    
    except json.JSONDecodeError as e:
        print(f"Failed to parse JSON. Error: {e.msg} at line {e.lineno}, column {e.colno}")
        return None

    



# TODO: Run it on ALL SIX letters; collect results into a pandas DataFrame (one row per
#   letter) and display it.

results = []

for i in range(len(LETTERS)):
    r = extract_fields(LETTERS[f'L00{i+1}'])
    results.append(r)


df = pd.DataFrame(results)
df
````

### Part 4.1 — Extraction accuracy against gold labels

````python
# TODO: For the three letters in GOLD, compare your extracted DataFrame to the gold values
#   field by field. Compute per-field accuracy across the three letters
#   (name matching can be case-insensitive; numbers must match exactly).

import re

# Label the extracted rows by letter ID (worth doing back in Part 3.2 as well)
df.index = list(LETTERS.keys())

FIELDS = list(GOLD["L001"].keys())
STOPWORDS = {"a", "an", "and", "the", "of", "for", "to", "into", "or", "new", "some"}


def tokens(text):
    """Lowercase content words, punctuation stripped."""
    return {w for w in re.findall(r"[a-z]+", str(text).lower()) if w not in STOPWORDS}


def is_null(value):
    return value is None or (pd.api.types.is_scalar(value) and pd.isna(value))


def purpose_overlap(got, want):
    """Fraction of the gold keywords that also appear in the extracted purpose.
    Prefix matching, so wash/washing and phone/phones still count as a hit."""
    g, w = tokens(got), tokens(want)
    if not w:
        return 0.0
    hits = sum(1 for t in w if any(t.startswith(u[:4]) or u.startswith(t[:4]) for u in g))
    return hits / len(w)


def matches(field, got, want):
    # a null is only correct when the gold value is null too
    if is_null(got) or is_null(want):
        return is_null(got) and is_null(want)

    if field == "applicant_name":
        return str(got).strip().lower() == str(want).strip().lower()

    if field == "purpose":
        # free text, so an exact match is impossible - accept >= 50% keyword overlap
        return purpose_overlap(got, want) >= 0.5

    if field == "has_collateral_or_guarantor":
        return bool(got) == bool(want)

    return float(got) == float(want)  # numeric fields must match exactly


# TODO: Display a small table: rows = fields, columns = L001 / L003 / L006 / accuracy.

rows = []
for field in FIELDS:
    row = {lid: matches(field, df.loc[lid].get(field), GOLD[lid][field]) for lid in GOLD}
    row["accuracy"] = sum(row.values()) / len(GOLD)
    rows.append(row)

accuracy_table = pd.DataFrame(rows, index=FIELDS)

correct = int(accuracy_table[list(GOLD)].values.sum())
total = len(FIELDS) * len(GOLD)
print(f"Overall extraction accuracy: {correct}/{total} fields correct ({correct / total:.1%})\n")

pretty = accuracy_table.copy()
for lid in GOLD:
    pretty[lid] = pretty[lid].map({True: "correct", False: "WRONG"})
pretty["accuracy"] = pretty["accuracy"].map("{:.0%}".format)
display(pretty)

# Every mismatch spelled out, so you can quote them in the Student Reasoning cell
print("\nMismatches:")
for field in FIELDS:
    for lid in GOLD:
        if not accuracy_table.loc[field, lid]:
            print(f"  {lid} / {field}: extracted {df.loc[lid].get(field)!r}  |  gold {GOLD[lid][field]!r}")
````

### Part 4.2 — Reliability: is the system consistent?

````python
# TODO: Run extract_fields() on letter L004 FIVE times at temperature=0 and FIVE times at
#   temperature=1.0.
run_1 = extract_fields(LETTERS['L004'])
run_2 = extract_fields(LETTERS['L004'])
run_3 = extract_fields(LETTERS['L004'])
run_4 = extract_fields(LETTERS['L004'])
run_5 = extract_fields(LETTERS['L004'])


run_6 = extract_fields(LETTERS['L004'], temperature=1.0)
run_7 = extract_fields(LETTERS['L004'], temperature=1.0)
run_8 = extract_fields(LETTERS['L004'], temperature=1.0)
run_9 = extract_fields(LETTERS['L004'], temperature=1.0)
run_10 = extract_fields(LETTERS['L004'], temperature=1.0)



# TODO: For each temperature, report how many of the 5 runs produced (a) valid JSON and
#   (b) identical values across runs. A simple approach: json.dumps(result, sort_keys=True)
#   and count unique strings.

temp0_runs = [run_1, run_2, run_3, run_4, run_5]
temp1_runs = [run_6, run_7, run_8, run_9, run_10]


def report_consistency(runs, temperature):
    # (a) how many runs returned parseable JSON, and
    # (b) how many produced identical values.

    valid = [r for r in runs if r is not None]   # extract_fields returns None when parsing fails

    # a run's "fingerprint" - sorted keys so field order never counts as a difference
    fingerprints = [json.dumps(r, sort_keys=True) for r in valid]
    unique = sorted(set(fingerprints))

    if not unique:
        verdict = "no valid output"
    elif len(unique) == 1:
        verdict = "all runs agreed"
    else:
        verdict = "the runs disagreed"

    print(f"----- Temp = {temperature} -----")
    print(f"valid JSON: {len(valid)}/{len(runs)} runs")
    print(f"unique results: {len(unique)}  ({verdict})")

    if valid:
        biggest = max(unique, key=fingerprints.count)
        print(f"largest identical group: {fingerprints.count(biggest)}/{len(runs)} runs")


    fields = sorted({k for r in valid for k in r})
    varied = {f: {json.dumps(r.get(f)) for r in valid} for f in fields}
    varied = {f: v for f, v in varied.items() if len(v) > 1}

    if varied:
        print("fields that varied:")
        for field, values in varied.items():
            print(f"{field}: {' | '.join(sorted(values))}")
    else:
        print("fields that varied: none")
    print()


report_consistency(temp0_runs, 0.0)
report_consistency(temp1_runs, 1.0)
````
